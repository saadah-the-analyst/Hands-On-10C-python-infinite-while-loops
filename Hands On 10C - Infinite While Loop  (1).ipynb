# Lesson 10 C - Infinite While Loop

# HANDS-ON 1: User Message Collector

```python
# Start an intentional infinite loop to collect continous user input
while True:
# prompt the user to enter a message or 'Q' to exit
    message = input("Enter a message or Q to quit: ")
    if message.upper() == "Q":           # Exit check: convert to uppercase so both 'q' and 'Q' quit the loop               
        break
        
    print(message)               # Display the entered message
```

# HANDS-ON 2: Simple Command-Line Support System

```python
# Start infinite loop for continous command input
while True:
# prompt the uuser to enter a command 
    command  = input("Enter a command: ")
    

#  check which command the user entered
    if command == "help":
        print("Available commands: help, status, quit, name, version")
    elif command == "status":
        print("System is running.")
    elif command ==  "name":
        print("Python Support System")
    elif command == "version":
        print("Python Support System v1.0")
    elif command == "quit":
        print("Closing system...")
        break         # Exit the loop and stop the program
    else:
        print("Unknown command.")
    
```

# HANDS-ON 3: Simple Data Entry System

```python
# Initialize a variabble outside the loop to keep track of total sales
total_sales = 0
# Start an intentional infinite loop to collect sales data
while True:
# Prompt the user for input and convert it to a number
    amount = int(input("Enter sales amount or 0 to finish:"))

    # check for the exit condition befolre processing the sales amount 
    if amount == 0:
        print("Data entry completed.")
        break                   # terminate the loop immediately
    
# Add the valid sales amount to the running total
    total_sales += amount

        # Display the valid sales record 
    print(f"Sales amount entered: {amount}")

# Display the grand total after the loop terminates
print(f"Total sales:{int(total_sales)}")
```

# BONUS HANDS-ON: The Controlled Loop Challenge

```python
 # initialize counter for valid entries
valid_entries = 0

while True:
# Prompt user, strip spaces and convert to lower case
    subject = input("What did you study today? ").strip().lower()

    # check for quit condition first
    if subject == "quit":
        print("Study session completed.")
        break

# Check valid subjects and update counter
    elif subject == "python":
        print("Great! keep practicing Python.")
        valid_entries += 1
    elif subject == "sql":
        print("Great! keep practicing SQL.")
        valid_entries += 1
    elif subject == "excel":
        print("Great! keep practicing Excel.")
        valid_entries += 1

        # Handle unrecognissed inputs
    else:
        print("Subject not recognized. Try again.")

# print the total count outside the loop
print(f"Valid study entries: {valid_entries}")
    
```
